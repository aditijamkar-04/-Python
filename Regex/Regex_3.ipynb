```python
import re

def valid_emails(emails):
    result = []
    pattern = r"\w+\d+\w+@\w+\.\w+"

    for email in emails:
        match = re.match(pattern, email)

        if match:
            result.append(True)
        else:
            result.append(False)

    return result


emails=["abc123@gmail.com","abc@gmail.com","abc123@hotmail.com","a@bc.com"]
valid_emails(emails)
```
