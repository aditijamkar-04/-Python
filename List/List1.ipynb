```python
def my_length(lst):
    count=0
    for i in lst:
        count+=1
    return count
```

```python
lst=[3, 4, 5, 100, 7, 8]
my_length(lst)
```

```python
def my_length_recursion(lst):
    if lst==[]:
        return 0 
    else:
        return 1+my_length_recursion(lst[1:])
```

```python
lst=[3, 4, 5, 100, 7, 8]
my_length_recursion(lst)
```

```python
def min_max(lst):
    max_ele=min_ele=lst[0]
    for i in range(len(lst)):
        if lst[i]>max_ele:
            max_ele=lst[i] 
        if lst[i]<min_ele:
            min_ele=lst[i]
    return min_ele,max_ele
```

```python
lst=[3, 4, 5, 100, 7, 8]
min_max(lst)
```
