Q: What is the difference between remove() and pop()?    
    
Answer: remove() deletes an element using its value,    
while pop() deletes an element using its index and also returns the removed element.

```python
numbers = [10, 200, 30, 40]

print(len(numbers))       # 4
print(max(numbers))       # 40
print(min(numbers))       # 10
print(sum(numbers))       # 100
print(sorted(numbers))    # [10, 20, 30, 40]
print(numbers)
```

```python
numbers = [1, 0, 0, 0]           # any()
print(any(numbers))              # Returns True if at least one value is True.
                                 # Because 1 is considered True.
```

```python
numbers = [2, 4, 7, 8]                    # Because 7 is odd.
print(any(i % 2 != 0 for i in numbers))    # ANY = at least one
```

```python
numbers = [2, 4, 6, 8]                      # all()
print(all(i % 2 == 0 for i in numbers))     # Returns True if all values are True.
                                            # All numbers are even.
```

```python
numbers = [2, 4, 7, 8]                        # Because 7 is not even.
print(all(i % 2 == 0 for i in numbers))       # ALL = every one
```

```python
fruits = ["Apple", "Banana", "Mango"]            # enumerate()
for index, fruit in enumerate(fruits):           # Used when you want both the index and value while looping through a list.
    print(index, fruit)
```

```python
fruits = ["Apple", "Banana", "Mango"]            # Without enumerate()
for i in range(len(fruits)):
    print(i, fruits[i])
```

```python
names = ["Aditi", "Rahul", "Priya"]              # zip()
marks = [85, 90, 78]                             # Combines elements from two or more lists.
for name, mark in zip(names, marks):
    print(name, mark)
```

```python
a = [1, 2, 3]
b = [4, 5, 6]

print(list(zip(a, b)))
```

```python
numbers = [1, 2, 3]               # append()
numbers.append(4)                 # Adds one element at the end of a list.
numbers
```

```python
numbers = [1, 2, 3]              # extend()
numbers.extend([4, 5, 6])         # Adds multiple elements to the end of a list.
print(numbers)
```

```python
numbers = [1, 2, 4]                   # insert()
numbers.insert(2, 3)                  # Adds an element at a specific position.
print(numbers)                        # insert(index, value)
```

```python
numbers = [10, 20, 30, 20]         # Remove()
numbers.remove(20)                 # Removes the first occurrence of a value.
print(numbers)
```

```python
numbers = [10, 20, 30]           # pop()
numbers.pop()                    # Removes and returns an element.
print(numbers)
```

```python
numbers = [10, 20, 30]           #  Remove using index
numbers.pop(1)
print(numbers)
```

```python
numbers = [1, 2, 3, 4]          # clear()
numbers.clear()                 # Removes all elements from the list.
print(numbers)
```

```python
numbers = [10, 20, 30, 20]       # index()
print(numbers.index(20))         # Returns the index/position of the first occurrence.
                                 # indexing starts from 0.
```

```python
numbers = [10, 20, 30, 40]
numbers[2] = 100
print(numbers)
```

```python
numbers = [10, 20, 20, 30, 20]     # count()
print(numbers.count(20))           # Counts how many times a value occurs.
```

```python
numbers = [5, 2, 8, 1, 3]           # sort()
sorted(numbers)                      # Sorts the list in ascending order.
```

```python
numbers = [5, 2, 8, 1, 3]           # sort()
numbers.sort()                      # Sorts the list in ascending order.
print(numbers)
```

```python
numbers = [5, 2, 8, 1, 3]          # Descending
numbers.sort(reverse=True)
print(numbers)
```

```python
numbers = [1, 2, 3]                # copy()
new_numbers = numbers.copy()       # Creates a copy of the list.
print(new_numbers)
```
