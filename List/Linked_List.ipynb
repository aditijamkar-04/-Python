```python
def create_LL(lst):
    LL={
        "data":lst[0],
        "link":None
    }
    tmp=LL 
    for i in range(1,len(lst)):
        new_node={
            "data":lst[i],
            "link":None
        }
        tmp["link"]=new_node 
        tmp=new_node 
    return LL
```

```python
LL=[10,20,30,40,50]
create_LL(lst)
```

```python
def count_nodes_LL(LL):
    count=0
    tmp=LL
    while tmp!=None:
        count+=1 
        tmp=tmp["link"]
    return count
```

```python
LL={'data': 10,
 'link': {'data': 20,
  'link': {'data': 30,
   'link': {'data': 40,
    'link': {'data': 50, 'link': {'data': 60, 'link': None}}}}}}
count_nodes_LL(LL)
```

```python
def adding_node_at_beggining(LL,node):
    new_node={
        "data":node,
        "link":LL
    }
    return new_node
```

```python
LL={'data': 10,
 'link': {'data': 20,
  'link': {'data': 30,
   'link': {'data': 40,
    'link': {'data': 50, 'link': {'data': 60, 'link': None}}}}}}
node=1000
adding_node_at_beggining(LL,node)
```

```python
def add_node_At_end(LL,node):
    new_node={
        "data":node,
        "link":None}
    tmp=LL
    while tmp["link"]!=None:
        tmp=tmp["link"]
    tmp["link"]=new_node
    return LL
```

```python
LL={'data': 10,
 'link': {'data': 20,
  'link': {'data': 30,
   'link': {'data': 40,
    'link': {'data': 50, 'link': {'data': 60, 'link': None}}}}}}
add_node_At_end(LL,2000)
```
