```python
def append_without_plus(lst,element):
    new_lst=[0]*(len(lst)+1)
    for i in range(len(lst)):
        new_lst[i]=lst[i]
    new_lst[-1]=element 
    return new_lst
```

```python
lst=[3,4,5,6,7,8,9]
element=1000
append_without_plus(lst,element)
```

```python
def append_recursive(lst,element):
    if len(lst)==0:
        return [element]
    else:
        return [lst[0]]+append_recursive(lst[1:],element)
```

```python
lst=[3,4,5,6,7,8,9]
element=1000
append_recursive(lst,element)
```
