```python
numbers = [10, 20, 10, 30, 20, 40]
res=[]
duplicates = []

for x in numbers:
    if numbers.count(x) > 1 and x not in duplicates:
        duplicates.append(x)
    else:
        res.append(x)
print(duplicates)
print(res)
```

```python
n = [-10, 20, 0, 40]
for i in n:
    if i>0:
        print("positive")
    elif i<0:
        print("negative")
    else:
        print("zero")
```

```python
# Find how many elements are greater than 20.
numbers = [10, 25, 30, 15, 40]
count = 0
for x in numbers:
    if x > 20:
        count += 1
print(count)
```

```python
# Find the sum of even numbers.
numbers = [10, 15, 20, 25, 30]
total = 0
for x in numbers:
    if x % 2 == 0:
        total += x
print(total)
```

```python
# Find the sum of odd numbers.
numbers = [10, 15, 20, 25, 30]
total = 0
for x in numbers:
    if x % 2 != 0:
        total += x
print(total)
```

```python
# Count even numbers.
numbers = [10, 15, 20, 25, 30]
count = 0
for x in numbers:
    if x % 2 == 0:
        count += 1
print(count)
```

```python
# Count odd numbers.
numbers = [10, 15, 20, 25, 30]
count = 0
for x in numbers:
    if x % 2 != 0:
        count += 1
print(count)
```

```python
# Create a list containing only even numbers.
numbers = [10, 15, 20, 25, 30]
even = []
for x in numbers:
    if x % 2 == 0:
        even.append(x)
print(even)
```

```python
# Create a list containing only odd numbers.
numbers = [10, 15, 20, 25, 30]
odd = []
for x in numbers:
    if x % 2 != 0:
        odd.append(x)
print(odd)
```

```python
# Find duplicate elements.
numbers = [10, 20, 10, 30, 20, 40]
seen = []
duplicates = []
for x in numbers:
    if x in seen:
        duplicates.append(x)
    else:
        seen.append(x)
print(list(duplicates))
```

```python
# Remove duplicate elements while maintaining order.
numbers = [10, 20, 10, 30, 20, 40]
lst = []
for x in numbers:
    if x not in lst:
        lst.append(x)
print(lst)
```

```python
# Find the frequency of every element.
numbers = [10, 20, 10, 30, 20, 10]
freq= []
for x in numbers:
    if x not in freq:
        print(x, numbers.count(x))
        freq.append(x)
```

```python
# Find the largest even number.
numbers = [15, 20, 35, 40, 25, 50]
even = []
for x in numbers:
    if x % 2 == 0:
        even.append(x)
print(max(even))
```

```python
# Find the smalllest odd number.
numbers = [15, 20, 35, 40, 25, 50]
odd = []
for x in numbers:
    if x % 2 != 0:
        odd.append(x)
print(min(odd))
```

```python
# Add 100 after every element greater than 20.
numbers = [10, 25, 15, 30]
res = []
for x in numbers:
    res.append(x)
    if x > 20:
        res.append(100)
print(res)
```
