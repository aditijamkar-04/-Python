```python
# Move all zeros to the end.
n = [0, 10, 0, 20, 30, 0]
res=[]
for i in n:
    if i!=0:
        res.append(i)
print(res)
for i in n:
    if i ==0:
        res.append(i)
print(res)

```

```python
# Find common elements between two lists.
a=[1,2,3,4,5]
b=[4,5,6,7]
res=[]
for i in a:
    for j in b:
        if i==j:
            res.append(i)
print(res)
```

```python
# Find elements present in the first list but not the second.
a=[1,2,3,4,5]
b=[4,5,6,7]
res=[]
for i in a:
    if i not in b:
        res.append(i)
print(res)
```

```python
# Combine two lists.
a=[1,2,3,4,5]
b=[4,5,6,7]
res=[]
for i in a:
    if i not in b:
        res.append(i)
print(res)
for j in b:
    res.append(j)
print(res)
```

```python
# Find whether a list contains duplicate elements.
numbers = [10, 20, 30, 20]
duplicate = False
for x in numbers:
    if numbers.count(x) > 1:
        duplicate = True
print(duplicate)
```

```python
# Find the element with the highest frequency.
n=[1,2,3,45,45,1,2,45]
ele=n[0]
for i in n:
    if n.count(i)>n.count(ele):
        ele=i
ele
```

```python
# Find pairs whose sum is 10.
numbers = [2, 8, 3, 7, 5, 5]
pairs = []
for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] + numbers[j] == 10:
            pairs.append([numbers[i], numbers[j]])
print(pairs)
```
