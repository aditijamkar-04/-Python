```python
# Find factorial
def factorial(n):
    i=1
    fact=1
    while i<=n:
        fact*=i
        i+=1
    return fact
factorial(4)
```

```python
# Reverse a number
def reverse(n):
    rev=0
    while n>0:
        digit=n%10
        rev=rev*10+digit
        n//=10
    return rev
reverse(13435)
```

```python
# Count vowels
def count(n):
    vowels=0
    for i in str(n):
        if i in "aeiou":
            vowels+=1
    print(vowels)
count("kjedhfgkaegwf")
```

```python
def count(n):
    i=0
    vowels=0
    while i<len(n) :
        if n[i] in "aeiou":
            vowels+=1
        i+=1
    print(vowels)
count("saaaagsfyfdjsaa")
```
