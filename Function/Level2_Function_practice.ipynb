```python
# Check positive or negative
def postive_or_negative(n):
    if n>0:
        return "positive"
    elif n<0:
        return "negative"
    else:
        return "zero"
postive_or_negative(10)
```

```python
# Check even or odd
def odd_or_even(n):
    if n%2==0:
        return "even"
    else:
        return "odd"
odd_or_even(10)
```

```python
# Find greater of two numbers
def greater(a,b):
    if a>b:
        return a
    else:
        return b
greater(10,50)
```

```python
# Check voting eligibility
def eligibility(age):
    if age<18:
        return "not eligible"
    else:
        return "eligible"
eligibility(20)
```

```python
# Check whether number is divisible by 5
def divisibility(n):
    if n%5==0:
        return "divisible by 5"
    else:
        return "not divisible by 5"
divisibility(3)
```

```python
# Check leap year
def leap_year(n):
    if n%100==0 or n%4==0 and n%400:
        return "leap year"
    else:
        return "not leap year"
leap_year(2024) 
```

```python
# Find largest of three numbers
def largest(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c
largest(10,30,9)
```

```python
# Check vowel or consonant
def vowels_or_consonants(str):
    if str in "aeiou":
        return "Vowels"
    else:
        return "Consonants"
vowels_or_consonants("d")
```

```python
# Check palindrome string
def palindrome(str):
    if str==str[::-1]:
        return True
    else:
        return False
palindrome("dfsgdaf")
```

```python
palindrome("madam")
```

```python
# Check Armstrong number without loops
def armstrong(n):
    a = n // 100 
    b = (n // 10) % 10
    c = n % 10

    if a**3 + b**3 + c**3 == n:
        return "Armstrong"
    else:
        return "Not Armstrong"
armstrong(153)
```

```python
# Check whether a number is divisible by 3 and 5
def divisibility(n):
    if n%5==0 and n%3==0:
        return "divisible by 5 and 3"
    else:
        return "not divisible by 5 and 3"
divisibility(15)
```

```python
# Electricity bill calculator
def electricity_bill(units):
    if units <= 100:
        return units * 5
    elif units <= 200:
        return 100 * 5 + (units - 100) * 7
    else:
        return 100 * 5 + 100 * 7 + (units - 200) * 10
electricity_bill(250)
```
