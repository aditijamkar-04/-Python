```python
#WAp to c heck the func find the mean and standard daviation of athe first n numbers
def find_mean_and_sd(n):
    i=1
    sum=0
    variance=0
    while i<=n:
        sum+=i
        i+=1
    mean=sum/n
    for i in range(1,n+1):
        variance+=(i-mean)**2
    variance/=n
    sd=(variance)**0.5
    return mean,sd
```

```python
n=int(input('enter a number:'))
mean,sd=find_mean_and_sd(n)
```

```python
#WAP sum of squres of n no. using function
def square(n):
    sos=0
    for i in range(1,n+1):
        sos+=i**2
    return sos
```

```python
n=int(input('enter a number:'))
square(n)
```

```python
# WAp to check a particular np. is a perfect no. or not
def perfect(n):
    divisor=0
    for i in range(1,n):
        if n%i==0:
            divisor+=i
    if n==divisor:
        return True
    else:
        return False
```

```python
n=int(input('enter a number:'))
perfect(n)
```

```python
#WAP to print all the divisor of a no.
def all(n):
    for i in range(1,n):
        if n%i==0:
            print(i)
```

```python
n=int(input('enter a number:'))
all(n)
```
