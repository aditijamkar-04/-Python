```python
# Q1. Add two numbers
def two_no(a,b):
    print(a+b)
two_no(10,20)
```

```python
# Q2. Find square of a number
def square(n):
    print(n**2)
square(20)
```

```python
# Q3. Find area of a rectangle
def rectangle(l,b):
    return l*b
rectangle(10,20)
```

```python
# Q4. Convert Celsius to Fahrenheit --->F=(C×9/5)+32
def celcius_to_fahrenheit(c):
    return (c*9/5)+32
celcius_to_fahrenheit(37)
```

```python
# Q5. Find average of three numbers
def avg(a,b,c):
    return (a+b+c)/3
avg(10,20,30)
```

```python
# Q6. Calculate simple interest 
def SI(p,r,t):
    return (p*r*t)/100
SI(5000,4,2)
```

```python
# Q7. Calculate power
def power(a,b):
    return a**b
power(9,2)
```

```python
# Q8. Calculate rectangle perimeter
def perimeter_of_rectangle(l,b):
    return (l+b)*2
perimeter_of_rectangle(10,20)
```

```python
# Q9. Calculate circle area
def area_of_circle(r):
    return 3.14*r*r
area_of_circle(5)
```

```python
# Q10. Return first and last character
def first_and_last_char(str):
    return str[0],str[-1]
str="hdgqwkhvk"
first_and_last_char(str)
```

```python
# Q11. Reverse a string
def rev(str):
    return str[::-1]
rev("python")
```

```python
# Q12. Find string length
def length(str):
    return len(str)
length("python")
```

```python
# Q13. Convert string to uppercase
def uppercase(str):
    return str.upper()
uppercase("python")
```

```python
# Q14. Find remainder
def remainder(a,b):
    return a%b
remainder(10,3)
```

```python
# Q15.Remove spaces from string
def remove_spaces(str):
    return str.replace(" ", "")
remove_spaces("Python Programming")
```
