```python
t = (10, 20, 30, 20, 40)
print(len(t))                     # len() — Count elements
print(max(t))                     # max() — Find largest value
print(min(t))                     # min() — Find smallest value
print(sum(t))                     # sum() — Add all values
print(sorted(t))                  # sorted() — Sort the tuple     # sorted() returns a list, not a tuple.
print(tuple(sorted(t)))
print(any(t))                    
print(all(t))                     # all() — All values are True
print(t.count(20))                # count() — Count an element            # Because 20 appears twice.
print(t.index(30))                # index() — Find position
print(t.index(20))                # Even though 20 also exists at index 3. it will still give 1
t.index(value, start)
```

```python
t = (10, 20, 0, 40)               # All values are non-zero, so all are considered True.
print(all(t))                     # all()
                                # any() → At least ONE is True
                                # all() → EVERY value is True
```

```python
t = (0, 0, 10, 0)                # any() — At least one value is True          
print(any(t))                    # 0 is considered False, while any non-zero number is considered True.
```

```python
t = (10, 20, 30, 40)                     # enumerate() — Index + value
for index, value in enumerate(t):
    print(index, value)
```

```python
names = ("Amit", "Rahul", "Sneha")          # zip() — Combine two tuples
marks = (80, 90, 85)
for name, mark in zip(names, marks):
    print(name, mark)
```
