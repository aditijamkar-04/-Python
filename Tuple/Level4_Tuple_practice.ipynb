```python
# Check whether at least one element is negative
t = (5, 8, -2, 10, 7)
print(any(x<0 for x in t))
```

```python
# Find the frequency of the most repeated element
t = (2, 4, 2, 7, 4, 2, 9, 4, 4)
print(max(t.count(x)for x in set(t)))
```

```python
# Check whether a tuple contains duplicates
t = (1, 2, 3, 4, 3)
print(len(t)!=len(set(t)))
```

```python
# Check whether all elements are unique
t = (10, 20, 30, 40)
print(len(t)==len(set(t)))
```

```python
# Find the index of the minimum value
t = (25, 10, 40, 5, 30)
print(t.index(min(t)))
```
