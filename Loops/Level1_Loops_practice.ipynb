```python
# Q1. Print numbers from 1 to 10.
i=1
while i<=10:
    print(i)
    i+=1
```

```python
for i in range(1,11):
    print(i)
```

```python
# Q2. Print numbers from 10 to 1.
i=10
while i>0:
    print(i)
    i-=1
```

```python
for i in range(10,0,-1):
    print(i)
```

```python
# Q3. Print all even numbers from 1 to 20.
i=1
while i<=20:
    if i%2==0:
        print(i)
    i+=1
```

```python
for i in range(1,21):
    if i%2==0:
        print(i)
```

```python
# Q4. Print all odd numbers from 1 to 20.
i=1
while i<=20:
    if i%2!=0:
        print(i)
    i+=1
```

```python
for i in range(1,21):
    if i&1!=0:
        print(i)
```

```python
# Q5. Print the multiplication table of 5.
i=5
while i<=100:
    if i%5==0:
        print(i)
    i+=1
```

```python
for i in range(1, 11):
    print(5 * i)
```

```python
# Q6. Print the multiplication table of a user-entered number.
n=int(input("n:"))
i=1
while i<=100:
    if i%n==0:
        print(i)
    i+=1
```

```python
n=int(input("n:"))
for i in range(1,11):
        print(i*n)
```

```python
# Q7. Find the sum of numbers from 1 to 10.
i=1
sum=0
while i<=10:
    sum+=i
    i+=1
print(sum)
```

```python
# Q10. Count the number of characters in a string using a loop.
n=input("n:")
count=0
for i in n:
    count+=1
print(count)
```
