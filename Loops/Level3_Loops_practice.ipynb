```python
# Q21. Find the smallest digit in a number.
n=int(input("n:"))
min_digit=9
while n>0:
    digit=n%10
    if min_digit>digit:
        min_digit=digit
    n//=10
print(min_digit)
```

```python
n=int(input("n:"))
min_digit=9
for i in range(len(str(n))):
    digit=n%10
    if min_digit>digit:
        min_digit=digit
    n//=10
print(min_digit)
```

```python
# Q22. Count even and odd digits in a number.
n = int(input("Enter a number: "))

even = 0
odd = 0

while n > 0:
    digit = n % 10

    if digit % 2 == 0:
        even += 1
    else:
        odd += 1

    n //= 10

print("Even:", even)
print("Odd:", odd)
```

```python
# Q23. Check whether a number is prime.
n = int(input("Enter a number: "))
count = 0
for i in range(1, n + 1):
    if n % i == 0:
        count += 1
if count == 2:
    print("Prime")
else:
    print("Not Prime")
```

```python
# Q24. Print all prime numbers from 1 to 100.
n = 1
while n <= 100:
    i = 1
    count = 0
    while i <= n:
        if n % i == 0:
            count += 1
        i += 1
    if count == 2:
        print(n)
    n += 1
```

```python
# Q25. Find the sum of prime numbers from 1 to 100.
n = 2
sum = 0
while n <= 100:  
    i = 2  
    factors = 0  
    while i < n:  
        if n % i == 0:
            factors += 1
        i += 1
    if factors == 0:  
        sum += n
    n += 1
print(sum)
```

```python
# Q27. Check whether a number belongs to the Fibonacci series.
n = int(input("Enter a number: "))
a = 0
b = 1
while a < n:
    a, b = b, a + b
if a == n:
    print("Fibonacci number")
else:
    print("Not a Fibonacci number")
```

```python
# Q30. Check whether a number is an Armstrong number.
n = int(input("Enter a number: "))
sum = 0
tmp = n          
while tmp > 0:
    digit = tmp % 10
    sum += digit ** 3
    tmp //= 10
if sum == n:
    print("Armstrong")
else:
    print("Not an Armstrong")
print(n)
```

```python
n = int(input("Enter a number: "))
temp = n
digits = len(str(n))
sum = 0
for i in str(n):
    sum += int(i) ** digits
if sum == temp:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```
