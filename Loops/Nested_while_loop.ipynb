```python
# WAP to find all the even numbers sum from n numbers
number=int(input("enter the number:"))
i=1
even_sum=0
while i<=number:
    if i&1==0:      # i%2==0
        even_sum+=i
    i+=1
print(even_sum)
```

```python
# WAP find the diff in the sum of even and odd of the n numbers
number=int(input("enter any number:"))
i=1
even_sum=0
odd_sum=0
while i<=number:
    if i&1==0:
        even_sum+=i
    else:
        odd_sum+=i
    i+=1
print(even_sum)
print(odd_sum)
diff=even_sum-odd_sum
if diff<0:
    print(-diff)
else:
    print(diff) 
```

```python
# WAP to find the nCr
n=int(input("n:"))
r=int(input("r:"))
n_fact=1
r_fact=1
n_minus_r=n-r
n_minus_r_fact=1
i=1
j=1
k=1
while i<=n:
    n_fact*=i
    i+=1
while j<=r:
    r_fact*=j
    j+=1
while k<=n_minus_r:
    n_minus_r_fact*=k
    k+=1
nCr=n_fact/(r_fact*n_minus_r_fact)
print(nCr)
```

```python
#WAP to find the sum of digits of a numbers   
#123%10---->3..........123//10----->12......12%10--->2....12//10--->1.....1%10---->1.....1//10--->0

n=int(input("n:"))
digit=0
sum=0
while n>0:
    digit=n%10
    sum+=digit
    n//=10
print(sum)
```

```python
#WAP to find the greatest digit in a number
# n=12642     o/p=6
n=int(input("n:"))
max_digit = 0
while n > 0:
    digit = n % 10        # extract last digit
    if digit > max_digit: # compare with current max
        max_digit = digit
    n //= 10            # remove last digit
print("The greatest digit is:", max_digit)
```

```python
# WAP to check if a number is an Armstrong number
n = int(input("Enter a number: "))
sum = 0
tmp = n          #temporary variable

while tmp > 0:
    digit = tmp % 10
    sum += digit ** 3
    tmp //= 10

if sum == n:
    print("Armstrong")
else:
    print("Not an Armstrong")
print(n)
```

```python
#WAP to print reverse of a number
# n=123  o/p=321     logic==> 123=1*100+2*10+3*1 ---->123
n=int(input("enter a number"))
rev=0
while n>0:
    digit=n%10
    rev=digit+rev*10
    n//=10
print (rev)
```

```python
# WAP to check a particular number is pallindrome or not 
number=int(input("enter any number"))
tmp=number
rev=0
while tmp>0:
    digit=tmp%10 
    rev=digit+rev*10 
    tmp//=10 
if number==rev:
    print("pallindrome")
else:
    print("not pallindrome")
```

```python
# WAP to check a particular number is prime number or not
n = int(input("Enter a number: "))

if n <= 1:
    print("Not prime")
else:
    i = 2
    while i < n:
        if n % i == 0:
            print("Not prime")
            break
        i += 1
    else:
        print("Prime")
```

```python
#WAP to print the all armstrong number in specific range
# constraints--> only for 3 digit
# Armstrong numbers in a given range (only 3-digit)
a = int(input("Lower limit: "))
b = int(input("Upper limit: "))
num = a
while num <= b:
    if 100 <= num <= 999:   # only 3-digit numbers
        s = 0
        temp = num
        while temp > 0:
            digit = temp % 10
            s += digit ** 3
            temp //= 10
        if s == num:
            print(num)
    num += 1
```
