```python
#print first n numbers 
for i in range(0,10,1):
    print(i)
```

```python
for i in range(10):
    for j in range(10):
        if i<j and i%2==0:
            for k in range(5):
                print(i+j-3>2)    #O(n^2*m) ....... sc--->O(1) 
```
