```python
# Q11. Count vowels in a string.
n=input("n:")
i=0
count=0
while i<len(n):
    ch=n[i]
    if ch in "aeiou":
        count+=1
    i+=1
print(count)
```

```python
n=input("n:")
count=0
for ch in n:
    if ch in "aeiou":
        count+=1
print(count)
```

```python
# Q13. Find the factorial of a number.
n=int(input("n:"))
i=1
fact=1
while i<=n:
    fact*=i
    i+=1
print(fact)
```

```python
# Q18. Reverse a number.
n=int(input("n:"))
rev=0
while n>0:
    digit=n%10
    rev=rev*10+digit
    n//=10
print(rev)
```

```python
# Q19. Check whether a number is a palindrome.
# A palindrome number is an integer that reads the same both forward and backward
n=int(input("n:"))
original=n
rev=0
while n>0:
    digit=n%10
    rev=rev*10+digit
    n//=10
print(rev)
if original==rev:
        print("Pallidrome")
else:
        print("Not Pallidrome")
```

```python
n = int(input("n: "))
rev = 0
for i in range(len(str(n))):
    digit = n % 10
    rev = rev * 10 + digit
    n //= 10
print(rev)
if original==rev:
        print("Pallidrome")
else:
        print("Not Pallidrome")
```

```python
# Q20. Find the largest digit in a number.
n=int(input("n:"))
max_digit=0
while n>0:
    digit=n%10
    if max_digit<digit:
        max_digit=digit
    n//=10
print(max_digit)
```
