```python
# Q37. Print a square pattern:
n=int(input("n:"))
i=1
while i<=n:
    j=1
    while j<=n:
        print("*",end="")
        j+=1
    print()
    i+=1
```

```python
n=int(input("n:"))
for i in range(1,n+1):
    for j in range(n-i):
        print(" ", end="")

    for j in range(i):
        print("*",end="")
    print()
```
