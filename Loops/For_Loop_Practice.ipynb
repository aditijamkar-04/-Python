# For Loop

```python
for i in range(2):
    for j in range(3):
        print(i, j)
```

```python
for i in range(3):
    for j in range(i):
        print("*", end="")
    print()
```

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()
```

```python
for i in range(2):
    for j in range(2):
        print(i + j)
```

```python
count = 0
for i in range(3):
    for j in range(2):
        count += 1
print(count)
```

```python
for i in range(3):
    for j in range(3):
        if i == j:
            print(i, j)
```

```python
for i in range(2):
    for j in range(2):
        print(i * j)
```

```python
for i in range(1, 4):
    for j in range(i):
        print(i, end="")
    print()
```

```python
for i in range(3):
    print(i)
    for j in range(2):
        print(j)
```

```python
s = 0
for i in range(3):
    for j in range(3):
        s += i + j
print(s)
```
