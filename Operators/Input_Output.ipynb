```python
# write a program to find the area of a triangle by using herons formula
a=int(input ("enter first side of a triangle:"))
b=int(input ("enter second side of a triangle:"))
c=int(input ("enter third side of a triangle:"))
s=(a+b+c)/2
area=(s*(s-a)*(s-b)*(s-c))**0.5
print(area)
```

```python
# write a program to find the area of a equilateral triangle
side=int(input("enter the side of equilateral triangle:"))
area=((3**0.5)/4)*(side**2)
print(area)
```
