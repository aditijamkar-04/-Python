```python
# Find the longest word.
s=input()        # i/p:  "Find the longest word in this sentence"
words=s.split()
longest_word = words[0]   # assume first word is longest
for w in words:
    if len(w) > len(longest_word):          # longest_word= max(words, key=len)
        longest_word = w
print(longest_word)
```

```python
# Find the shortest word.
s=input()        # i/p:  "Find the longest word in this sentence"
words=s.split()
shortest_word = words[0]   
for w in words:
    if len(w) < len(shortest_word):
        shortest_word = w
print(shortest_word)
```

```python
# Count each character.
s=input()
seen=""
for ch in s:
    if ch not in seen: 
        print(ch,s.count(ch))
        seen+=ch
```

```python
# Count vowels and consonants.
s=input().lower()
for ch in s:
    if ch in "aeiou":
        print(ch,s.count(ch))
    if ch not in "aeiou":
        print(ch,s.count(ch))
```

```python
s=input().lower()
vowels=0
consonants=0
for i in s:
    if i in "aeiou":
        vowels+=1
    else:
        consonants+=1
print(vowels)
print(consonants)
```

```python
# Count alphabets, digits, and special characters.
s=input()
alphabets=0
digits=0
special_char=0
for i in s:
    if i.isalpha():
        alphabets+=1
    elif i.isdigit():
        digits+=1
    else:
        special_char+=1
print(alphabets)
print(digits)
print(special_char)
```

```python
# Find the most frequent character.
s=input()
max_char=0
a=""
for ch in s:
    count=s.count(ch)
    if count>max_char:
        max_char=count
        a=ch
print(a)
print(max_char)
```

```python
# Find the least frequent character.
s=input()
min_char=len(s)
r=""
for ch in s:
    count=s.count(ch)
    if count<min_char:
        min_char=count
        r=ch
print(r)
print(min_char)
```

```python
# Remove duplicate words from a sentence.
s=input()
word=s.split()
d=[]
for i in word:
    if i not in d:
        d.append(i)
print(" ".join(d))
```

```python
# Find the second longest word.
s=input().split()
s.sort(key=len,reverse=True)
l=[]
for i in s:
    if i not in l:
        l.append(i)
print((l[1]))
```

```python
# Count words that start with a vowel.
s=input().split()
count=0
for i in s:
    if i.lower().startswith(('a','e','i','o','u')):
        count+=1
print(count)
```

```python
# Check password validity.
password=input()
if len(password)>=8 and any(i.isalpha() for i in password) and any(i.isdigit() for i in password):
    print("valid")
else:
    print("not")
```

```python
s = input()
words = s.split()
longest = ""
for word in words:
    if word == word[::-1] and len(word) > len(longest):
        longest = word
print(longest)
```
