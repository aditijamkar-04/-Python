```python
s="Python is easy"
words=s.split()
r=[]
for i in words:
    r.append(i[::-1])
print(" ".join(r))
```

```python
# Find how many digits are present using count().
s = input()
count = 0
for digit in "0123456789":
    count += s.count(digit)

print(count)
```

```python
# Replace vowels with *.
s = input()
s = s.replace("a", "*")
s = s.replace("e", "*")
s = s.replace("i", "*")
s = s.replace("o", "*")
s = s.replace("u", "*")
print(s)
```
