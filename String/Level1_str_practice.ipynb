```python
# Check Whether a Word Exists
text = input("Enter string: ")
word = input("Enter word: ")
if word in text:
    print("Found")
else:
    print("Not Found")
```

```python
# Reverse a String
str=input("n:")
print(str[::-1])
```

```python
# Check String Palindrome
text = input("Enter string: ")

if text[::1] == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# count vowels and consonants
text = input("Enter string: ")

vowels = 0
consonants = 0

for ch in text:
    if ch in "aeiouAEIOU":
        vowels += 1
    elif ch.isalpha():
        consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```
