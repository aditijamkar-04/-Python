```python
# point inside , outside or on the circle with the help of radius
radius=float(input("enter the radius:"))
x_cor=int(input("enter the x-coordinates of a point:"))
y_cor=int(input("enter the y-coordinates of a point:"))
distance=((x_cor**2)+(y_cor**2))**0.5
if distance < radius:
    print("point lies inside the circle")
if distance> radius:
     print("point lies outside the circle")
if distance == radius:
     print("point lies on the circle")
```

```python
# to check particular point lies in which quadrant of a 2D space
# Program to check which quadrant a point lies in

x = int(input("Enter x-coordinate: "))
y = int(input("Enter y-coordinate: "))

if x > 0 and y > 0:
    print("Point lies in Quadrant I")
elif x < 0 and y > 0:
    print("Point lies in Quadrant II")
elif x < 0 and y < 0:
    print("Point lies in Quadrant III")
elif x > 0 and y < 0:
    print("Point lies in Quadrant IV")
elif x == 0 and y == 0:
    print("Point lies at the Origin")
elif x == 0:
    print("Point lies on Y-axis")
elif y == 0:
    print("Point lies on X-axis")
```

```python
# to check a kth bit of a number is set or not from right side
number=int(input("enter any number:"))
k=int(input("enter k value"))
if (number >> k) & 1 == 1:
    print("True")
else:
    print("false")
```

```python
number = int(input("Enter any number: "))
k = int(input("Enter k value (from left side, starting at 1): "))

total_bits = number.bit_length()

if (number >> (total_bits - k)) & 1 == 1:
    print("True - kth bit from left is set")
else:
    print("False - kth bit from left is not set")
```
