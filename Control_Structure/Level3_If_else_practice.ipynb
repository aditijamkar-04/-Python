```python
# Q1. Find the largest among three numbers.
a=int(input("a:"))
b=int(input("b:"))
c=int(input("c:"))
if a >= b and a >= c:
    print(a, "is largest")
elif b >= a and b >= c:
    print(b, "is largest")
else:
    print(c, "is largest")
```

```python
# Q2. Check whether a year is a leap year.
year=int(input("Enter the year:"))
if year%100!=0  and  year%4==0 or year%400==0:
    print("Leap year")
else:
    print("Not leap year")
```

```python
# Q3. Check whether a triangle is valid.
a=int(input("a:"))
b=int(input("b:"))
c=int(input("c:"))
if a + b > c and a + c > b and b + c > a:
    print("Valid triangle")
else:
    print("Invalid triangle")
```

```python
# Q4. Determine the type of triangle.
a=int(input("a:"))
b=int(input("b:"))
c=int(input("c:"))
if a==b and b==c and a==c:
    print("All equal → Equilateral")
elif a==b or b==c or a==c:
    print("Two equal → Isosceles")
else:
    print("All different → Scalene")
```

```python
# Q5. Calculate electricity bill.
units = int(input("Enter units: "))
if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 7
else:
    bill = units * 10
print("Bill =", bill)
```

```python
# Q6. Calculate discount based on purchase amount.
amount = float(input("Enter purchase amount: "))
if amount >= 5000:
    discount = amount * 0.20
elif amount >= 3000:
    discount = amount * 0.15
elif amount >= 1000:
    discount = amount * 0.10
else:
    discount = 0

final_amount = amount - discount
print("Discount =", discount)
print("Final amount =", final_amount)
```

```python
# Q7. Check whether a number is a 1-digit, 2-digit, 3-digit, or 4+ digit number.
num = abs(int(input("Enter number: ")))     # abs()---> absolute value.
if num <= 9:
    print("1-digit number")
elif num <= 99:
    print("2-digit number")
elif num <= 999:
    print("3-digit number")
else:
    print("4 or more digit number")
```

```python
# Q8. Calculate student result.
Eng=int(input("Eng Marks:"))
Phy=int(input("Phy Marks:"))
Maths=int(input("Maths Marks:"))
if Eng>=40 and Phy>= 40 and Maths>=40:
    print("Passed")
else:
    print("Failed")
    
total = Eng+ Phy+Maths
percentage = total / 3
print("Total:",total)
print("Percentage:",percentage)
```

```python
# Q10. Check a number's divisibility.
n=int(input("n:"))
if n%2==0 and n%3==0:
    print("Divisible by both 2 and 3")
elif n%2==0:
    print("Divisible only by 2")
elif n%3==0:
    print("Divisible only by 3")
else:
    print("Divisible by neither")
```
