```python
# Q1. ATM withdrawal validation.
balance = float(input("Enter balance: "))
amount = int(input("Enter withdrawal amount: "))

if amount <= 0:
    print("Invalid amount")
elif amount % 100 != 0:
    print("Amount must be a multiple of 100")
elif amount > balance:
    print("Insufficient balance")
else:
    balance -= amount
    print("Withdrawal successful")
    print("Remaining balance:", balance)
```

```python
# Q2. Calculate income tax based on salary.
salary=float(input("salary:"))
if salary<=250000:
    tax = 0
elif salary<=500000:
    tax=salary*(5/100)
elif salary<=1000000:
    tax=salary*(20/100)
else:
    tax=salary*(30/100)
print("Tax:", tax)
print("remaining salary:",salary-tax)
```

```python
# Q3. Movie ticket pricing.
age=int(input("age:"))
student=input("person is student (yes/no):")
if age<5:
    price=0
elif age<=12:
    price=100
elif age<=59:
    price=200
else:
    price=120
if student == "yes" and price > 0:
    price -= 30

print("Ticket price =", price)
```

```python
# Q4. Employee bonus calculation.
salary=float(input("salary:"))
experience=int(input("Experience:"))
if salary>=50000 and experience>=5:
    bonus=salary*(20/100)
elif salary>=30000 and experience>=3:
    bonus=salary*(10/100)
else:
    bonus=salary*(5/100)

print("Bonus:",bonus)
print("total salary:",salary+bonus)
```

```python
# Q5. Password strength checker.
password = input("Enter password: ")
length = len(password)
if length < 6:
    print("Weak password")
elif length < 10:
    print("Medium password")
else:
    print("Strong password")
```

```python
# Q6. Shopping eligibility.
purchase=int(input("Purchase:"))
premium = input("Are you a premium member? (yes/no): ")
if purchase>=1000 or premium == "yes":
    delivery_charges=0
else:
    delivery_charges=100

print("Total Purchase:",purchase+delivery_charges)
```

```python
# Q7. Find the middle number among three different numbers.
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))
if (a > b and a < c) or (a < b and a > c):
    print("Middle =", a)
elif (b > a and b < c) or (b < a and b > c):
    print("Middle =", b)
else:
    print("Middle =", c)
```
