```python
#Q1. Check whether a number is positive or negative.
num = int(input("Enter a number: "))
if num > 0:
    print("Positive")
else:
    print("Negative")
```

```python
# Q2. Check whether a number is even or odd.
num = int(input("Enter a number: "))
if num%2== 0:
    print("Even")
else:
    print("Odd")
```
