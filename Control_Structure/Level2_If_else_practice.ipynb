```python
# Q1. Check whether a number is positive, negative, or zero.
n = int(input("Enter a number: "))
if n<0:
    print("Negative")
elif n>0:
    print("Positive")
else:
    print("zero")
```

```python
# Q2. Display grade based on marks.
n = int(input("Enter a number: "))
if n>=90:
    print("Grade A")
elif n>=75:
    print("Grade B")
elif n>=60:
    print("Grade C")
elif n>=40:
    print("Grade D")
else:
    print("Grade F")
```

```python
# Q3. Find the greater of two numbers.
a = int(input("Enter a number: "))
b = int(input("Enter b number: "))
if a>b:
    print("a is greater")
elif b > a:
    print("b is greater")
else:
    print("Both are equal")
```

```python
# Q4. Check age category.
n = int(input("Enter a number: "))
if n<=12:
    print("Child")
elif n<=19:
    print("Teenager")
elif n<=59:
    print("Adult")
else:
    print("Senior Citizen")
```

```python
# Q5. Create a simple calculator.
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print(a + b)
elif operator == "-":
    print(a - b)
elif operator == "*":
    print(a * b)
elif operator == "/":
    if b != 0:
        print(a / b)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

```python
# Q6. Check whether a character is a vowel or consonant.
ch = input("Enter a char: ")
if ch == "a" or ch == "e" or ch == "i" or ch == "o" or ch == "u":
    print("Vowels")
else:
    print("Consonants")
```

```python
# Q8. Check whether a number is divisible by both 3 and 5.
num = int(input("Enter a number: "))
if num % 3 == 0 and num % 5 == 0:
    print("Divisible by both 3 and 5")
else:
    print("Not divisible by both")
```
